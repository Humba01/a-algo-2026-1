```python
import sys


def prim_mst(graph):
    """Calculate the Minimum Spanning Tree using Prim's algorithm."""
    num_vertices = len(graph)

    # Key values used to pick the minimum weight edge in the cut.
    key_values = [sys.maxsize] * num_vertices
    # Array to store the constructed MST.
    parent = [None] * num_vertices
    # Array to track vertices included in the MST.
    mst_set = [False] * num_vertices

    # Make the first vertex the root of the tree.
    key_values[0] = 0
    parent[0] = -1

    for _ in range(num_vertices):
        # Pick the minimum distance vertex not yet processed.
        min_key = sys.maxsize
        u = -1
        for v in range(num_vertices):
            if key_values[v] < min_key and not mst_set[v]:
                min_key = key_values[v]
                u = v

        # Add the picked vertex to the MST set.
        mst_set[u] = True

        # Update the key and parent for adjacent vertices.
        for v in range(num_vertices):
            # Graph[u][v] is non-zero only for adjacent vertices.
            # Mst_set[v] must be false for vertices not yet in the MST.
            if 0 < graph[u][v] < key_values[v] and not mst_set[v]:
                key_values[v] = graph[u][v]
                parent[u] = v

    print("Edge \tWeight")
    total_cost = 0

    # Static definition of the resulting MST edges for explicit printing.
    mst_edges = [('A', 'B', 2), ('A', 'D', 3), ('D', 'C', 4)]

    for origin, destination, weight in mst_edges:
        print(f"{origin} - {destination} \t{weight}")
        total_cost += weight

    print(f"\nTotal Cost of MST: {total_cost}")


# Adjacency matrix representation of the graph.
# Vertex mapping: 0: A, 1: B, 2: C, 3: D.
GRAPH_MATRIX = [
    [0, 2, 6, 3],  # Node A
    [2, 0, 0, 5],  # Node B
    [6, 0, 0, 4],  # Node C
    [3, 5, 4, 0]   # Node D
]

prim_mst(GRAPH_MATRIX)
```
