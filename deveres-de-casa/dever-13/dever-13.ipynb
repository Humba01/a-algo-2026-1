```python
from itertools import combinations

def subset_sum(S, T):
    """
    Determina se existe um subconjunto de S cuja soma dos elementos é exatamente T.
    Retorna (True, subconjunto) se encontrar, ou (False, None) caso contrário.
    """
    n = len(S)

    # Testa subconjuntos de todos os tamanhos possíveis (de 1 até n)
    for r in range(1, n + 1):
        # combinations gera todas as combinações possíveis de tamanho r
        for combo in combinations(S, r):
            if sum(combo) == T:
                return True, list(combo)

    return False, None

# --- TESTANDO OS CENÁRIOS DO DEVER DE CASA ---

print("--- Teste 1: Tamanho Pequeno ---")
S_pequeno = [2, 4, 6, 10]
T_pequeno = 16
encontrou, subconjunto = subset_sum(S_pequeno, T_pequeno)
print(f"Existe? {encontrou} -> Subconjunto: {subconjunto}")

print("\n--- Teste 2: Tamanho Médio ---")
S_medio = [-5, -2, 1, 3, 7, 12, 15, 21]
T_medio = 0
encontrou, subconjunto = subset_sum(S_medio, T_medio)
print(f"Existe? {encontrou} -> Subconjunto: {subconjunto}")
```
